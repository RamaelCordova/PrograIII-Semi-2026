Calcular el sueldo de un empleado. hrs * valor de horas * 30 dias las deducciones: ISSS, AFP, ISR (Ver tabla del MH)

```python
calcular = lambda horas, valor_hora: (
    float(horas) * float(valor_hora) * 30,  # sueldo base
    (lambda sueldo_base: (
        sueldo_base - sueldo_base*0.03 - sueldo_base*0.0725,  # sueldo gravado
        (lambda sueldo_gravado: (
            0 if sueldo_gravado <= 550 else
            (sueldo_gravado - 550)*0.10 + 17.67 if sueldo_gravado <= 895.24 else
            (sueldo_gravado - 895.24)*0.20 + 60.00 if sueldo_gravado <= 2038.10 else
            (sueldo_gravado - 2038.10)*0.30 + 288.57
        ))(sueldo_base - sueldo_base*0.03 - sueldo_base*0.0725)
    ))(float(horas) * float(valor_hora) * 30)
)

h = input("Horas trabajadas por día: ")
v = input("Valor de la hora: ")
sueldo_base, (sueldo_gravado, isr) = calcular(h, v)
neto = sueldo_gravado - isr

print("Sueldo base:", sueldo_base)
print("Sueldo gravado:", sueldo_gravado)
print("ISR:", isr)
print("Sueldo total:", neto)
```

```python
#FILTER
#dada una serie numero dame solo lso que son pares
serie = [2,5,6,8,9,10,12]

pares = filter(lambda num: num%2==0, serie)
impares = filter(lambda num: num%2!=0, serie)
print("Original: ", serie)
print("Pares: ", list(pares))
print("Impares: ", list(impares))
```

```python
#reduce
#Cuanto es el acumulado de los numeros pares de una serie
from functools import reduce

numeros = [1,2,8,4,9,6]
pares = filter(lambda num: num%2==0, numeros)

print("Numeros:" , numeros)
#print("Pares: ", list(pares))

sumaPares = reduce(lambda acum, num: num+acum, pares, 0)
print("Suma de pares: ", sumaPares)
```

Determinar si un número es par, impar o primo

```python

#El numero primo solo tiene dos divisores posibles: 1 y él mismo

numero = int(input("Introduce un número: "))

if numero % 2 == 0:
    print("El número es par")
else:
    print("El número es impar")

primo = True
if numero <= 1:
    primo = False
else:
    for i in range(2, int(numero**0.5) + 1):
        if numero % i == 0:
            primo = False
            break

if primo:
    print("El número es primo")
else:
    print("El número no es primo")
```

```python
print("==========================================")
print("          CONVERSOR DE TIEMPO")
print("==========================================")

while True:

    # UNIDAD DE ORIGEN
    print("\nSeleccione la unidad de ORIGEN:")
    print("1. Milisegundos")
    print("2. Segundos")
    print("3. Minutos")
    print("4. Horas")
    print("5. Días")
    print("6. Semanas")
    print("7. Meses")
    print("8. Años")
    print("9. Décadas")
    print("10. Siglos")

    origen = int(input("\nSeleccione la unidad de origen: "))

    # UNIDAD DE DESTINO
    print("\nSeleccione la unidad de DESTINO:")
    print("1. Milisegundos")
    print("2. Segundos")
    print("3. Minutos")
    print("4. Horas")
    print("5. Días")
    print("6. Semanas")
    print("7. Meses")
    print("8. Años")
    print("9. Décadas")
    print("10. Siglos")

    destino = int(input("\nSeleccione la unidad de destino: "))

    # CANTIDAD
    cantidad = float(input("\nIngrese la cantidad que desea convertir: "))

    # ==========================================
    # CONVERTIR LA UNIDAD DE ORIGEN A SEGUNDOS
    # ==========================================

    if origen == 1:
        segundos = cantidad / 1000

    elif origen == 2:
        segundos = cantidad

    elif origen == 3:
        segundos = cantidad * 60

    elif origen == 4:
        segundos = cantidad * 3600

    elif origen == 5:
        segundos = cantidad * 86400

    elif origen == 6:
        segundos = cantidad * 604800

    elif origen == 7:
        segundos = cantidad * 2592000

    elif origen == 8:
        segundos = cantidad * 31536000

    elif origen == 9:
        segundos = cantidad * 315360000

    elif origen == 10:
        segundos = cantidad * 3153600000

    else:
        print("\nUnidad de origen no válida.")
        continue

    # ==========================================
    # CONVERTIR SEGUNDOS A LA UNIDAD DE DESTINO
    # ==========================================

    if destino == 1:
        resultado = segundos * 1000
        unidad = "milisegundos"

    elif destino == 2:
        resultado = segundos
        unidad = "segundos"

    elif destino == 3:
        resultado = segundos / 60
        unidad = "minutos"

    elif destino == 4:
        resultado = segundos / 3600
        unidad = "horas"

    elif destino == 5:
        resultado = segundos / 86400
        unidad = "días"

    elif destino == 6:
        resultado = segundos / 604800
        unidad = "semanas"

    elif destino == 7:
        resultado = segundos / 2592000
        unidad = "meses"

    elif destino == 8:
        resultado = segundos / 31536000
        unidad = "años"

    elif destino == 9:
        resultado = segundos / 315360000
        unidad = "décadas"

    elif destino == 10:
        resultado = segundos / 3153600000
        unidad = "siglos"

    else:
        print("\nUnidad de destino no válida.")
        continue

    # ==========================================
    # MOSTRAR RESULTADO
    # ==========================================

    print("\n==========================================")
    print("              RESULTADO")
    print("==========================================")
    print(cantidad, "unidades =", resultado, unidad)
    print("==========================================")

    # ==========================================
    # CONTINUAR O TERMINAR
    # ==========================================

    print("\n¿Qué desea hacer?")
    print("11. Realizar otra conversión")
    print("12. Detener la ejecución")

    opcion = int(input("\nSeleccione una opción: "))

    if opcion == 11:
        print("\nIniciando una nueva conversión...")
        continue

    elif opcion == 12:
        print("\n==========================================")
        print("          PROGRAMA FINALIZADO")
        print("==========================================")
        break

    else:
        print("\nOpción no válida.")
        print("El programa finalizará.")
        break
```
