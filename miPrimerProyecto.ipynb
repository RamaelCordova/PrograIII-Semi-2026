Calcular el sueldo de un empleado. hrs * valor de horas * 30 dias las deducciones: ISSS, AFP, ISR (Ver tabla del MH)

```python
calcular = lambda horas, valor_hora: (
    float(horas) * float(valor_hora) * 30,  # sueldo base
    (lambda sueldo_base: (
        sueldo_base - sueldo_base*0.03 - sueldo_base*0.0725,  # sueldo gravado
        (lambda sueldo_gravado: (
            0 if sueldo_gravado <= 550 else
            (sueldo_gravado - 550)*0.10 + 17.67 if sueldo_gravado <= 895.24 else
            (sueldo_gravado - 895.24)*0.20 + 60.00 if sueldo_gravado <= 2038.10 else
            (sueldo_gravado - 2038.10)*0.30 + 288.57
        ))(sueldo_base - sueldo_base*0.03 - sueldo_base*0.0725)
    ))(float(horas) * float(valor_hora) * 30)
)

h = input("Horas trabajadas por día: ")
v = input("Valor de la hora: ")
sueldo_base, (sueldo_gravado, isr) = calcular(h, v)
neto = sueldo_gravado - isr

print("Sueldo base:", sueldo_base)
print("Sueldo gravado:", sueldo_gravado)
print("ISR:", isr)
print("Sueldo total:", neto)
```

FUNCIONES PRECONSTRUIDAS

```python
#MAP
semana = ["domingo", "lunes", "Martes", "Miercoles", "jueves", "Viernes", "sabado"]
print(semana,"Primer dia de la semana: ", semana[0])

primeraLetraDia = map(lambda dia: dia[0], semana)
print(list(primeraLetraDia))

primeraLetraMayusDia = map(lambda dia: dia[0].upper(), semana)
print(list(primeraLetraMayusDia))
```

```python
#Todos los numeros de la matriz elevarlo a un exponente
numeros = [2,4,6,7,9]
exponente = int(input("Exponente: "))
potencia = map(lambda num: num**exponente, numeros)
print(list(potencia))
```

```python
#FILTER
#dada una serie numero dame solo lso que son pares
serie = [2,5,6,8,9,10,12]

pares = filter(lambda num: num%2==0, serie)
impares = filter(lambda num: num%2!=0, serie)
print("Original: ", serie)
print("Pares: ", list(pares))
print("Impares: ", list(impares))
```

```python
#Hacer en python el siguiente ejercicio, haciendo uso de filter y lambda.
#La notas introducidas por el usuario clasificarlas en aprobadas y reprobadas
notas = input("Notas: ") #5,6,1,8,9
notas = notas.split(",")
aprobados = filter(lambda nota: float(nota)>=6, notas)
reprobados = filter(lambda nota: float(nota)<6, notas)
print("Notas: ", notas)
print("Aprobados: ", list(aprobados))
print("Reprobados: ", list(reprobados))
```

```python
#reduce
#Cuanto es el acumulado de los numeros pares de una serie
from functools import reduce

numeros = [1,2,8,4,9,6]
pares = filter(lambda num: num%2==0, numeros)

print("Numeros:" , numeros)
#print("Pares: ", list(pares))

sumaPares = reduce(lambda acum, num: num+acum, pares, 0)
print("Suma de pares: ", sumaPares)
```
